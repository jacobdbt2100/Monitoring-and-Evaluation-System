## 1. Imports

```python
# ------------------------------------------------------------
# Import Required Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import random
import uuid

from datetime import timedelta
```

## 2. Load Farmer Registration Data

```python
# ------------------------------------------------------------
# Load Farmer Registration Dataset
# ------------------------------------------------------------
# This dataset is used as the source for generating
# Training Attendance records.
#
# Existing fields such as:
# - form.case.@case_id
# - username
# - hq_user
#
# are copied from this dataset to preserve the
# relationship between both datasets.
# ------------------------------------------------------------

farmer_df = pd.read_csv("C:/Users/HP PC/Documents/jacob_git_projects/Monitoring-and-Evaluation-System/dataset/02_cleaned/"
                        "farmer_registration_clean.csv")

print(farmer_df.shape)

farmer_df.head()
```

## 3. Helper Function

```python
# ------------------------------------------------------------
# Helper Function
# ------------------------------------------------------------

def generate_form_id():
    return str(uuid.uuid4())
```

## 4. Training Options

```python
# ------------------------------------------------------------
# Training Attendance Options
# ------------------------------------------------------------

training_topics = [
    "good_agronomic_practices",
    "soil_fertility_management",
    "pest_and_disease_management",
    "irrigation_and_water_management",
    "livestock_management",
    "record_keeping",
    "other"
]

venues = [
    "Community Hall",
    "Primary School",
    "Farmer Cooperative Centre",
    "Village Square",
    "Extension Office"
]

trainers = [
    "Trainer 1",
    "Trainer 2",
    "Trainer 3",
    "Trainer 4",
    "Trainer 5"
]

partial_reasons = [
    "arrived_late",
    "left_early",
    "personal_emergency",
    "other"
]
```

## 5. Generate Training Attendance Data

```python
# ------------------------------------------------------------
# Generate Training Attendance Records
# ------------------------------------------------------------
# One training attendance record is generated for most farmers.
#
# Existing fields copied from Farmer Registration:
# - form.case.@case_id (Join Key)
# - username
# - hq_user
#
# These values already exist and should not be randomly
# generated.
# ------------------------------------------------------------

training_records = []

for _, farmer in farmer_df.iterrows():

    # --------------------------------------------------------
    # Existing farmer information
    # --------------------------------------------------------

    case_id = farmer["form.case.@case_id"]
    username = farmer["username"]
    hq_user = farmer["hq_user"]

    # --------------------------------------------------------
    # About 90% of farmers attend at least one training.
    # --------------------------------------------------------

    if random.random() > 0.90:
        continue

    form_id = generate_form_id()

    registration_date = pd.to_datetime(
        farmer["completed_time"]
    )

    training_datetime = registration_date + timedelta(
        days=random.randint(5, 120),
        hours=random.randint(8, 15),
        minutes=random.randint(0, 59)
    )

    started_time = training_datetime - timedelta(
        minutes=random.randint(2, 8)
    )

    completed_time = training_datetime

    received_on = completed_time + timedelta(
        seconds=random.randint(5, 20)
    )

    # --------------------------------------------------------
    # Training Topic
    # --------------------------------------------------------

    topic = random.choice(training_topics)

    if topic == "other":
        specify_other = random.choice([
            "Climate Smart Agriculture",
            "Post Harvest Handling",
            "Financial Literacy"
        ])
    else:
        specify_other = "---"

    # --------------------------------------------------------
    # Attendance
    # --------------------------------------------------------

    attended = random.choices(
        ["yes", "no"],
        weights=[88, 12],
        k=1
    )[0]

    if attended == "yes":

        attendance_status = random.choices(
            [
                "full_attendance",
                "partial_attendance"
            ],
            weights=[80, 20],
            k=1
        )[0]

        if attendance_status == "partial_attendance":

            partial_reason = random.choice(
                partial_reasons
            )

        else:

            partial_reason = "---"

    else:

        attendance_status = "---"
        partial_reason = "---"

    # --------------------------------------------------------
    # Save Record
    # --------------------------------------------------------

    training_records.append({

        "formid":
            form_id,

        "form.section_1_training_information.training_date":
            completed_time.strftime("%Y-%m-%d"),

        "form.section_1_training_information.training_topic.choice":
            topic,

        "form.section_1_training_information.training_topic.specify_other":
            specify_other,

        "form.section_1_training_information.training_venue":
            random.choice(venues),

        "form.section_1_training_information.trainer_name":
            random.choice(trainers),

        "form.section_2_attendance.did_the_farmer_attend_the_training":
            attended,

        "form.section_2_attendance.attendance_status":
            attendance_status,

        "form.section_2_attendance.reason_for_partial_attendance":
            partial_reason,

        # Join Key
        "form.case.@case_id":
            case_id,

        "completed_time":
            completed_time,

        "started_time":
            started_time,

        "username":
            username,

        "received_on":
            received_on,

        "form_link":
            f"https://www.commcarehq.org/a/commcare-learn-ahmad/reports/form_data/{form_id}/",

        "hq_user":
            hq_user

    })
```

## 6. Create DataFrame

```python
# ------------------------------------------------------------
# Create Training Attendance DataFrame
# ------------------------------------------------------------

training_df = pd.DataFrame(
    training_records
)

training_df.insert(
    0,
    "number",
    range(len(training_df))
)

print(
    f"Training Records Generated: {len(training_df)}"
)

training_df.head()
```

## 7. Quick Quality Checks

```python
# ------------------------------------------------------------
# Quick Quality Checks
# ------------------------------------------------------------

print(training_df.shape)

print("\nUnique Farmers:")
print(
    training_df["form.case.@case_id"].nunique()
)

print("\nAttendance:")

print(
    training_df[
        "form.section_2_attendance.did_the_farmer_attend_the_training"
    ].value_counts()
)

print("\nAttendance Status:")

print(
    training_df[
        "form.section_2_attendance.attendance_status"
    ].value_counts()
)
```

## 8. Export CSV

```python
# ------------------------------------------------------------
# Export Training Attendance Dataset
# ------------------------------------------------------------

training_df.to_csv("C:/Users/HP PC/Documents/jacob_git_projects/Monitoring-and-Evaluation-System/dataset/01_raw/training_attendance_raw.csv",
                     index=False)

print("Training Attendance dataset exported successfully.")
```
