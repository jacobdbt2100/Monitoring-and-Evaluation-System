## Import Required Libraries

```python
import pandas as pd
import numpy as np
```

## Load Raw Dataset

```python
df = pd.read_csv("C:/Users/HP PC/Documents/jacob_git_projects/Monitoring-and-Evaluation-System/dataset/01_raw/training_attendance_raw.csv")

print(df.shape)

df.head()
```

```python
# ------------------------------------------------------------
# Replace CommCare placeholder values with missing values
# ------------------------------------------------------------
# CommCare uses "---" where a question was not applicable.
# Replace these placeholders with blank (NaN) values so that
# numeric, date, and boolean fields can be imported into
# PostgreSQL without data type errors.
# ------------------------------------------------------------

df.replace("---", np.nan, inplace=True)
```

## Data Validation Checks

```python
print("Dataset Shape")
print(df.shape)

print("\nDuplicate Rows")
print(df.duplicated().sum())

print("\nMissing Values")
print(df.isna().sum())

print("\nData Types")
print(df.dtypes)
```

## Export Clean Dataset

```python
# ------------------------------------------------------------
# No cleaning is required for this dataset.
# It has passed the validation checks.
# ------------------------------------------------------------

df.to_csv("C:/Users/HP PC/Documents/jacob_git_projects/Monitoring-and-Evaluation-System/dataset/02_cleaned/training_attendance_clean.csv",
          index=False)

print("Training Attendance dataset exported successfully.")
```
