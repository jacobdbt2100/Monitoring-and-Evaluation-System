## 1. Imports

```python
# ------------------------------------------------------------
# Import Required Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import random
import uuid

from datetime import timedelta
```

## 2. Load Farmer Registration Data

```python
# ------------------------------------------------------------
# Load Farmer Registration Dataset
# ------------------------------------------------------------
# Existing farmer information is copied from this dataset.
# ------------------------------------------------------------

farmer_df = pd.read_csv("C:/Users/HP PC/Documents/jacob_git_projects/Monitoring-and-Evaluation-System/dataset/02_cleaned/"
                        "farmer_registration_clean.csv")

print(farmer_df.shape)

farmer_df.head()
```

## 3. Helper Function

```python
# ------------------------------------------------------------
# Helper Function
# ------------------------------------------------------------

def generate_form_id():
    return str(uuid.uuid4())
```

## 4. Input Distribution Options

```python
# ------------------------------------------------------------
# Input Distribution Options
# ------------------------------------------------------------

locations = [
    "Community Hall",
    "Village Square",
    "Primary School",
    "Extension Office",
    "Farmer Cooperative Centre"
]
```

## 5. Generate Input Distribution Data

```python
# ------------------------------------------------------------
# Generate Input Distribution Records
# ------------------------------------------------------------
# Existing values copied from Farmer Registration:
#
# - form.case.@case_id (Join Key)
# - username
# - hq_user
#
# These should not be randomly generated.
# ------------------------------------------------------------

distribution_records = []

for _, farmer in farmer_df.iterrows():

    # --------------------------------------------------------
    # About 85% of farmers receive programme inputs.
    # --------------------------------------------------------

    if random.random() > 0.85:
        continue

    case_id = farmer["form.case.@case_id"]
    username = farmer["username"]
    hq_user = farmer["hq_user"]

    registration_date = pd.to_datetime(
        farmer["completed_time"]
    )

    distribution_datetime = registration_date + timedelta(
        days=random.randint(10, 120),
        hours=random.randint(8, 15),
        minutes=random.randint(0, 59)
    )

    started_time = distribution_datetime - timedelta(
        minutes=random.randint(2, 8)
    )

    completed_time = distribution_datetime

    received_on = completed_time + timedelta(
        seconds=random.randint(5, 20)
    )

    form_id = generate_form_id()

    # --------------------------------------------------------
    # Determine eligible inputs
    # --------------------------------------------------------

    available_inputs = []

    crops = str(
        farmer[
            "form.section_4_crops_cultivated.crops_cultivated_and_areas_hectares.choice"
        ]
    ).split()

    livestock = str(
        farmer[
            "form.section_5_livestock.livestock_and_number_kept_integer.choice"
        ]
    ).split()

    if "maize" in crops:
        available_inputs.append("maize_seed")

    if "rice" in crops:
        available_inputs.append("rice_seed")

    if "groundnuts" in crops:
        available_inputs.append("groundnut_seed")

    if len(crops) > 0:
        available_inputs.append("fertilizer")

    if "goats" in livestock:
        available_inputs.append("goats")

    if "rabbits" in livestock:
        available_inputs.append("rabbits")

    selected_inputs = random.sample(
        available_inputs,
        random.randint(
            1,
            len(available_inputs)
        )
    )

    input_type = " ".join(selected_inputs)

    maize_seed = "---"
    rice_seed = "---"
    groundnut_seed = "---"
    fertilizer = "---"
    goats = "---"
    rabbits = "---"

    if "maize_seed" in selected_inputs:
        maize_seed = round(
            random.uniform(2, 10),
            1
        )

    if "rice_seed" in selected_inputs:
        rice_seed = round(
            random.uniform(2, 10),
            1
        )

    if "groundnut_seed" in selected_inputs:
        groundnut_seed = round(
            random.uniform(2, 8),
            1
        )

    if "fertilizer" in selected_inputs:
        fertilizer = round(
            random.uniform(25, 100),
            1
        )

    if "goats" in selected_inputs:
        goats = random.randint(1, 3)

    if "rabbits" in selected_inputs:
        rabbits = random.randint(1, 5)

    signature = (
        "https://www.commcarehq.org/"
        "a/commcare-learn-ahmad/"
        "api/form_attachment/v1/"
        f"{form_id}/signature.png"
    )

    distribution_records.append({

        "formid":
            form_id,

        "form.distribution_date":
            completed_time.strftime("%Y-%m-%d"),

        "form.distribution_location":
            random.choice(locations),

        "form.input_type":
            input_type,

        "form.quantity_of_inputs_distributed.maize_seed_kg":
            maize_seed,

        "form.quantity_of_inputs_distributed.rice_seed_kg":
            rice_seed,

        "form.quantity_of_inputs_distributed.groundnut_seed_kg":
            groundnut_seed,

        "form.quantity_of_inputs_distributed.fertilizer_kg":
            fertilizer,

        "form.quantity_of_inputs_distributed.goats_count":
            goats,

        "form.quantity_of_inputs_distributed.rabbits_count":
            rabbits,

        "form.recipient_signature":
            signature,

        # Join Key
        "form.case.@case_id":
            case_id,

        "completed_time":
            completed_time,

        "started_time":
            started_time,

        "username":
            username,

        "received_on":
            received_on,

        "form_link":
            f"https://www.commcarehq.org/a/commcare-learn-ahmad/reports/form_data/{form_id}/",

        "hq_user":
            hq_user

    })
```

## 6. Create DataFrame

```python
# ------------------------------------------------------------
# Create Input Distribution DataFrame
# ------------------------------------------------------------

distribution_df = pd.DataFrame(
    distribution_records
)

distribution_df.insert(
    0,
    "number",
    range(len(distribution_df))
)

print(
    f"Input Distribution Records Generated: {len(distribution_df)}"
)

distribution_df.head()
```
