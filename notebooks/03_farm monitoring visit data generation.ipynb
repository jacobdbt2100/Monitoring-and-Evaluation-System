## 1. Import Required Libraries

```python
# ------------------------------------------------------------
# Import Required Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import random
import uuid
from datetime import datetime, timedelta
```

## 2. Load Farmer Registration Dataset

```python
# ------------------------------------------------------------
# Load Farmer Registration Dataset
# ------------------------------------------------------------
# Read the cleaned Farmer Registration dataset.
# This dataset provides the existing farmers that will be
# monitored. It also supplies the case_id used to link each
# monitoring visit to the correct farmer.
# ------------------------------------------------------------

farmer_df = pd.read_csv("C:/Users/HP PC/Documents/jacob_git_projects/Monitoring-and-Evaluation-System/dataset/02_cleaned/"
                        "farmer_registration_clean.csv")

print(f"Farmers loaded: {len(farmer_df)}")

farmer_df.head()
```

```python
print(farmer_df.shape)
```

## 3. Helper Functions

```python
# ------------------------------------------------------------
# Helper Functions
# ------------------------------------------------------------

def generate_form_id():
    return str(uuid.uuid4())


def random_datetime(start_date, end_date):

    delta = end_date - start_date

    random_seconds = random.randint(
        0,
        int(delta.total_seconds())
    )

    return start_date + timedelta(seconds=random_seconds)
```

## 4. Monitoring Visit Options

```python
# ------------------------------------------------------------
# Monitoring Visit Options
# ------------------------------------------------------------

crop_stages = [
    "land_preparation",
    "planting",
    "vegetative_growth",
    "flowering",
    "harvesting"
]

crop_conditions = [
    "good",
    "fair",
    "poor"
]

livestock_conditions = [
    "good",
    "fair",
    "poor"
]

adoption_status = [
    "yes",
    "no",
    "partially"
]

recommended_practices = [
    "improved_seed",
    "fertilizer_application",
    "proper_spacing",
    "timely_weeding",
    "pest_management",
    "improved_livestock_housing"
]

farm_challenges = [
    "pest_infestation",
    "crop_disease",
    "poor_rainfall",
    "flooding",
    "lack_of_fertilizer",
    "livestock_disease",
    "feed_shortage",
    "no_major_challenge"
]

comments = [
    "Farm progressing well.",
    "Crop growth is satisfactory.",
    "Minor challenges observed.",
    "Farmer following recommendations.",
    "Monitoring completed successfully."
]
```

## Steps 5 - 13

```python
## 5. Generate Monitoring Visits


# ------------------------------------------------------------
# Generate Monitoring Visits
# ------------------------------------------------------------
# Loop through each registered farmer and generate between
# one and three monitoring visits.
#
# Existing information such as case_id and username is copied
# from the Farmer Registration dataset instead of being
# randomly generated. This preserves the relationship between
# both datasets and allows them to be joined later using
# form.case.@case_id.

# ------------------------------------------------------------

monitoring_records = []

for _, farmer in farmer_df.iterrows():

    # Existing farmer information (copied)
    case_id = farmer["form.case.@case_id"]
    username = farmer["username"]
    hq_user = farmer["hq_user"]

    # Generate between 1 and 3 monitoring visits
    number_of_visits = random.choice([1, 2, 2, 2, 3])


## 6. Generate Visit Date and Metadata


    # --------------------------------------------------------
    # Generate each monitoring visit
    # --------------------------------------------------------

    for visit in range(number_of_visits):
        
        form_id = generate_form_id()

        # Registration date
        registration_date = pd.to_datetime(
            farmer["completed_time"]
        )

        # Monitoring visit happens after registration
        visit_datetime = registration_date + timedelta(
            days=random.randint(7, 150),
            hours=random.randint(8, 16),
            minutes=random.randint(0, 59)
        )

        started_time = visit_datetime - timedelta(
            minutes=random.randint(2, 8)
        )

        completed_time = visit_datetime

        received_on = completed_time + timedelta(
            seconds=random.randint(5, 30)
        )


## 7. Generate Crop Information


        # ----------------------------------------------------
        # Crop Information
        # ----------------------------------------------------
        # Only observe crops that were registered by
        # the farmer.
        # ----------------------------------------------------

        registered_crops = str(
            farmer[
                "form.section_4_crops_cultivated.crops_cultivated_and_areas_hectares.choice"
            ]
        ).split()

        observed_crops = random.sample(
            registered_crops,
            random.randint(
                1,
                len(registered_crops)
            )
        )

        observed_crops = " ".join(observed_crops)

        maize_condition = "---"
        rice_condition = "---"
        groundnuts_condition = "---"

        if "maize" in observed_crops:
            maize_condition = random.choice(crop_conditions)

        if "rice" in observed_crops:
            rice_condition = random.choice(crop_conditions)

        if "groundnuts" in observed_crops:
            groundnuts_condition = random.choice(crop_conditions)


## 8. Generate Livestock Information


        # ----------------------------------------------------
        # Livestock Information
        # ----------------------------------------------------
        # Only monitor livestock that were registered.
        # ----------------------------------------------------

        registered_livestock = str(
            farmer[
                "form.section_5_livestock.livestock_and_number_kept_integer.choice"
            ]
        ).split()

        if registered_livestock == ["---"]:

            observed_livestock = "---"

        else:

            observed_livestock = random.sample(
                registered_livestock,
                random.randint(
                    1,
                    len(registered_livestock)
                )
            )

            observed_livestock = " ".join(
                observed_livestock
            )

        goat_condition = "---"
        rabbit_condition = "---"

        if "goats" in observed_livestock:
            goat_condition = random.choice(
                livestock_conditions
            )

        if "rabbits" in observed_livestock:
            rabbit_condition = random.choice(
                livestock_conditions
            )


## 9. Adoption of Recommended Practices


        # ----------------------------------------------------
        # Adoption of Recommended Farming Practices
        # ----------------------------------------------------

        adoption = random.choices(
            adoption_status,
            weights=[60, 15, 25],
            k=1
        )[0]

        if adoption == "yes":

            adopted_practices = random.sample(
                recommended_practices,
                random.randint(2, len(recommended_practices))
            )

            adopted_practices = " ".join(adopted_practices)

        elif adoption == "partially":

            adopted_practices = random.sample(
                recommended_practices,
                random.randint(1, 2)
            )

            adopted_practices = " ".join(adopted_practices)

        else:

            adopted_practices = "---"


## 10. Farm Challenges


        # ----------------------------------------------------
        # Farm Challenges
        # ----------------------------------------------------

        if random.random() < 0.65:

            challenges = "no_major_challenge"

        else:

            number_of_challenges = random.randint(1, 3)

            available_challenges = [
                challenge
                for challenge in farm_challenges
                if challenge != "no_major_challenge"
            ]

            challenges = random.sample(
                available_challenges,
                number_of_challenges
            )

            challenges = " ".join(challenges)

        comment = random.choice(comments)


## 11. Monitoring Visit GPS


        # ----------------------------------------------------
        # Monitoring Visit GPS
        # ----------------------------------------------------
        # Most monitoring visits should be very close to the
        # original registration GPS.
        #
        # A small percentage are intentionally generated far
        # away to support GPS validation during the
        # Data Quality Assessment stage.
        # ----------------------------------------------------

        registration_gps = farmer[
            "form.section_6_farm_location.farm_gps_location"
        ]

        latitude, longitude, altitude, accuracy = map(
            float,
            registration_gps.split()
        )

        if random.random() < 0.08:

            latitude += random.uniform(0.02, 0.05)
            longitude += random.uniform(0.02, 0.05)

        else:

            latitude += random.uniform(-0.0005, 0.0005)
            longitude += random.uniform(-0.0005, 0.0005)

        monitoring_gps = (
            f"{latitude:.7f} "
            f"{longitude:.7f} "
            f"{altitude:.1f} "
            f"{accuracy:.1f}"
        )


## 12. Photograph URL


        # ----------------------------------------------------
        # Monitoring Photograph
        # ----------------------------------------------------

        photo_number = random.randint(
            1783000000000,
            1784999999999
        )

        photograph = (
            "https://www.commcarehq.org/"
            "a/commcare-learn-ahmad/"
            "api/form_attachment/v1/"
            f"{form_id}/{photo_number}.jpg"
        )


## 13. Save Record


        # ----------------------------------------------------
        # Save Monitoring Visit Record
        # ----------------------------------------------------

        monitoring_records.append({

            "formid": form_id,

            "form.section_1_visit_information.visit_date":
                completed_time.strftime("%Y-%m-%d"),

            "form.section_1_visit_information.crop_stage":
                random.choice(crop_stages),

            "form.section_1_visit_information.crops_observed_during_this_visit":
                observed_crops,

            "form.section_1_visit_information.select_crop_condition.crop_condition_maize":
                maize_condition,

            "form.section_1_visit_information.select_crop_condition.crop_condition_rice":
                rice_condition,

            "form.section_1_visit_information.select_crop_condition.crop_condition_groundnuts":
                groundnuts_condition,

            "form.section_2_livestock_monitoring.which_livestock_were_observed_during_this_visit":
                observed_livestock,

            "form.section_2_livestock_monitoring.select_livestock_condition.goat_condition":
                goat_condition,

            "form.section_2_livestock_monitoring.select_livestock_condition.rabbit_condition":
                rabbit_condition,

            "form.section_3_adoption_of_recommended_practices.has_the_farmer_adopted_the_recommended_farming_practices":
                adoption,

            "form.section_3_adoption_of_recommended_practices.which_recommended_practices_have_been_adopted":
                adopted_practices,

            "form.section_4_farm_challenges.what_challenges_were_observed_during_the_visit":
                challenges,

            "form.section_4_farm_challenges.additional_comments":
                comment,

            "form.section_4_farm_challenges.capture_farm_photograph":
                photograph,

            "form.section_5_farm_location.capture_monitoring_visit_gps":
                monitoring_gps,

            # Join Key
            "form.case.@case_id":
                case_id,

            "completed_time":
                completed_time,

            "started_time":
                started_time,

            "username":
                username,

            "received_on":
                received_on,

            "form_link":
                f"https://www.commcarehq.org/a/commcare-learn-ahmad/reports/form_data/{form_id}/",

            "hq_user":
                hq_user

        })
```

## 14. Create DataFrame

```python
# ------------------------------------------------------------
# Create Monitoring Visit DataFrame
# ------------------------------------------------------------

monitoring_df = pd.DataFrame(monitoring_records)

monitoring_df.insert(
    0,
    "number",
    range(len(monitoring_df))
)

print(f"Monitoring Visits Generated: {len(monitoring_df)}")

monitoring_df.head()
```
