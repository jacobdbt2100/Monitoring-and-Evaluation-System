## 1. Imports and Configuration

```python
# ============================================================
# FARMER REGISTRATION - SYNTHETIC DATA GENERATOR
# End-to-End M&E Agriculture Project
# ============================================================

import random
import uuid
from datetime import datetime, timedelta

import numpy as np
import pandas as pd

# Make results reproducible
random.seed(42)
np.random.seed(42)

# ============================================================
# CONFIGURATION
# ============================================================

NUMBER_OF_FARMERS = 1000

START_DATE = datetime(2026, 1, 1)
END_DATE = datetime(2026, 6, 30)

USERNAME = "tester"

FORM_ID_PREFIX = "farmer_registration"

OUTPUT_FILE = "farmer_registration_raw.csv"
```

## 2. Lookup Lists

```python
# ============================================================
# LOOKUP VALUES
# ============================================================

communities = [
    "community_01",
    "community_02",
    "community_03",
    "community_04",
    "community_05",
    "community_06",
    "community_07"
]

community_weights = [
    0.18,
    0.16,
    0.145,
    0.14,
    0.135,
    0.125,
    0.115
]

genders = ["male", "female"]
gender_weights = [0.56, 0.44]

occupations = [
    "crop_farming",
    "mixed_farming",
    "livestock_farming"
]

occupation_weights = [
    0.50,
    0.35,
    0.15
]

ownership = [
    "owned",
    "family_land",
    "leased"
]

ownership_weights = [
    0.60,
    0.25,
    0.15
]

irrigation = [
    "rain-fed",
    "borehole",
    "river_stream",
    "irrigation_scheme",
    "dam_reservoir",
    "other"
]

irrigation_weights = [
    0.65,
    0.15,
    0.08,
    0.05,
    0.04,
    0.03
]

other_irrigation = [
    "Rainwater Harvesting",
    "Community Well",
    "Spring Water"
]

crop_options = [
    "groundnuts",
    "maize",
    "rice"
]

livestock_options = [
    "goats",
    "rabbits"
]
```

## 3. Name Lists

```python
# ============================================================
# SYNTHETIC NAMES
# ============================================================

first_names = [
    "Amina","Grace","Fatima","Mary","Sarah","Rebecca",
    "John","Peter","James","Musa","Ibrahim","Daniel",
    "Joseph","David","Samuel","Ruth","Esther","Mercy",
    "Blessing","Hauwa","Binta","Ahmed","Yusuf","Suleiman",
    "Emmanuel","Paul","Joy","Deborah","Naomi","Elizabeth"
]

last_names = [
    "Yusuf","Musa","Bello","James","Peter","Isa",
    "Daniel","Garba","Mohammed","Ahmed","Johnson",
    "Samuel","David","Usman","Ali","Ibrahim",
    "Yakubu","Sani","Okafor","Eze","Ojo",
    "Adeyemi","Ogunleye","Balogun","Lawal",
    "Abdullahi","Haruna","Tanko","Michael","Philip"
]
```

## 4. Helper Functions

```python
# ============================================================
# HELPER FUNCTIONS
# ============================================================

def random_date(start, end):
    """Generate a random datetime."""
    delta = end - start
    return start + timedelta(days=random.randint(0, delta.days))


def random_dob():

    age_group = random.choices(
        ["18-29", "30-39", "40-49", "50-59", "60-75"],
        weights=[18,28,25,18,11]
    )[0]

    if age_group == "18-29":
        age = random.randint(18,29)

    elif age_group == "30-39":
        age = random.randint(30,39)

    elif age_group == "40-49":
        age = random.randint(40,49)

    elif age_group == "50-59":
        age = random.randint(50,59)

    else:
        age = random.randint(60,75)

    today = datetime(2026,7,1)

    birth = today - timedelta(days=age*365 + random.randint(0,364))

    return birth.date()


def farming_experience(age):

    maximum = max(1, age - 18)

    return random.randint(1, maximum)


def generate_phone():

    number = "080" + "".join(
        random.choices("0123456789", k=8)
    )

    return number


def generate_case_id():
    return str(uuid.uuid4())


def generate_form_id():
    return str(uuid.uuid4())
```

## 5. Generate Farmer Registration Records

```python
# ============================================================
# GENERATE SYNTHETIC FARMER REGISTRATION DATA
# ============================================================

farmers = []

for i in range(1, NUMBER_OF_FARMERS + 1):

    # -----------------------------
    # Community
    # -----------------------------
    community = random.choices(
        communities,
        weights=community_weights,
        k=1
    )[0]

    # -----------------------------
    # Name
    # -----------------------------
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)

    # -----------------------------
    # Gender
    # -----------------------------
    gender = random.choices(
        genders,
        weights=gender_weights,
        k=1
    )[0]

    # -----------------------------
    # Date of Birth
    # -----------------------------
    dob = random_dob()

    age = 2026 - dob.year

    # -----------------------------
    # National ID (Optional)
    # -----------------------------
    if random.random() < 0.65:
        national_id = str(random.randint(10000000000,99999999999))
    else:
        national_id = np.nan

    # -----------------------------
    # Phone Number
    # -----------------------------
    phone = generate_phone()

    # -----------------------------
    # Photograph
    # -----------------------------
    photograph = (
        "https://www.commcarehq.org/"
        f"photos/farmer_{i}.jpg"
    )

    # -----------------------------
    # Household
    # -----------------------------
    household_size = random.randint(3,8)

    if random.random() < 0.15:
        household_size = random.randint(1,2)

    if random.random() < 0.10:
        household_size = random.randint(9,12)

    occupation = random.choices(
        occupations,
        weights=occupation_weights,
        k=1
    )[0]

    experience = farming_experience(age)

    # -----------------------------
    # Farm Information
    # -----------------------------
    farm_ownership = random.choices(
        ownership,
        weights=ownership_weights,
        k=1
    )[0]

    plots = random.randint(1,3)

    if random.random() < 0.20:
        plots = random.randint(4,6)

    irrigation_source = random.choices(
        irrigation,
        weights=irrigation_weights,
        k=1
    )[0]

    specify_other = "---"

    if irrigation_source == "other":
        specify_other = random.choice(other_irrigation)

    # -----------------------------
    # Crops
    # -----------------------------
    selected_crops = random.sample(
        crop_options,
        random.randint(1,3)
    )

    crop_choice = " ".join(selected_crops)

    groundnuts_area = "---"
    maize_area = "---"
    rice_area = "---"

    if "groundnuts" in selected_crops:
        groundnuts_area = round(
            random.uniform(0.5,3.0),
            1
        )

    if "maize" in selected_crops:
        maize_area = round(
            random.uniform(0.5,6.5),
            1
        )

    if "rice" in selected_crops:
        rice_area = round(
            random.uniform(0.5,4.0),
            1
        )

    # -----------------------------
    # Livestock
    # -----------------------------
    livestock_choice = "---"

    goats = "---"
    rabbits = "---"

    livestock_selected = []

    if random.random() < 0.45:
        livestock_selected.append("goats")
        goats = random.randint(2,25)

    if random.random() < 0.20:
        livestock_selected.append("rabbits")
        rabbits = random.randint(2,40)

    if len(livestock_selected) > 0:
        livestock_choice = " ".join(livestock_selected)

    # -----------------------------
    # GPS
    # -----------------------------
    latitude = round(random.uniform(10.20,10.60),7)
    longitude = round(random.uniform(7.20,7.60),7)

    gps = f"{latitude} {longitude} 0.0 20.0"

    # -----------------------------
    # Consent
    # -----------------------------
    consent = "yes"

    if random.random() < 0.02:
        consent = "no"

    # -----------------------------
    # Metadata
    # -----------------------------
    case_id = generate_case_id()

    started = random_date(
        START_DATE,
        END_DATE
    )

    completed = started + timedelta(
        minutes=random.randint(2,8)
    )

    received = completed + timedelta(
        seconds=random.randint(5,20)
    )

    form_id = generate_form_id()

    form_link = (
        "https://www.commcarehq.org/"
        f"form/{form_id}"
    )

    # -----------------------------
    # Append Record
    # -----------------------------
    farmers.append({

        "number": i,
        "formid": form_id,

        "form.section_1_farmer_identification.community": community,
        "form.section_1_farmer_identification.first_name": first_name,
        "form.section_1_farmer_identification.last_name": last_name,
        "form.section_1_farmer_identification.gender": gender,
        "form.section_1_farmer_identification.date_of_birth": dob,
        "form.section_1_farmer_identification.national_id": national_id,
        "form.section_1_farmer_identification.phone_number": phone,
        "form.section_1_farmer_identification.farmer_photograph": photograph,

        "form.section_2_household_information.household_size": household_size,
        "form.section_2_household_information.primary_occupation": occupation,
        "form.section_2_household_information.farming_experience_years": experience,

        "form.section_3_farm_information.farm_ownership": farm_ownership,
        "form.section_3_farm_information.number_of_farm_plots": plots,
        "form.section_3_farm_information.primary_or_main_irrigation_source.choice": irrigation_source,
        "form.section_3_farm_information.primary_or_main_irrigation_source.specify_other": specify_other,

        "form.section_4_crops_cultivated.crops_cultivated_and_areas_hectares.choice": crop_choice,
        "form.section_4_crops_cultivated.crops_cultivated_and_areas_hectares.area_under_groundnuts_hectares": groundnuts_area,
        "form.section_4_crops_cultivated.crops_cultivated_and_areas_hectares.area_under_maize_hectares": maize_area,
        "form.section_4_crops_cultivated.crops_cultivated_and_areas_hectares.area_under_rice_hectares": rice_area,

        "form.section_5_livestock.livestock_and_number_kept_integer.choice": livestock_choice,
        "form.section_5_livestock.livestock_and_number_kept_integer.number_of_goats": goats,
        "form.section_5_livestock.livestock_and_number_kept_integer.number_of_rabbits": rabbits,

        "form.section_6_farm_location.farm_gps_location": gps,

        "form.section_7_consent.farmer_accepts_or_decline_to_participate_in_the_programme.choice": consent,

        "form.case.@case_id": case_id,

        "completed_time": completed,
        "started_time": started,
        "username": USERNAME,
        "received_on": received,
        "form_link": form_link,
        "hq_user": USERNAME

    })

# Create DataFrame

df = pd.DataFrame(farmers)

print(df.shape)

df.head()
```
